```python
import pandas as pd

def create_kulfan_coefficients_table():
    """
    Extract Kulfan coefficients from airfoil files and create a transposed LaTeX table
    """
    # Dictionary to store coefficients for each airfoil
    airfoil_data = {}
    
    # Process each airfoil file
    for f in files:
        if 'OSO_2025_WT2' in f:
            fnm = f.split('.')
            tau = fnm[0].split('_')[-1][1:]  # Extract thickness from filename
            if float(tau) >= 21:
                # Load the airfoil
                afl = Kulfan(TE_gap=te_gap_lookup[tau])
                afl.readFile(f)
                
                # Store coefficients
                airfoil_data[tau] = {
                    'thickness': float(tau),
                    'upper_coeffs': afl.upperCoefficients,
                    'lower_coeffs': afl.lowerCoefficients,
                    'te_gap': te_gap_lookup[tau]
                }
    
    # Sort by thickness
    sorted_taus = sorted(airfoil_data.keys(), key=lambda x: float(x))
    
    # Determine maximum number of coefficients
    max_upper = max(len(airfoil_data[tau]['upper_coeffs']) for tau in sorted_taus)
    max_lower = max(len(airfoil_data[tau]['lower_coeffs']) for tau in sorted_taus)
    
    # Create LaTeX table
    latex_content = []
    latex_content.append(r'\begin{table}[htbp]')
    latex_content.append(r'\centering')
    latex_content.append(r'\caption{Kulfan Coefficients for OSO 2025 WT2 Airfoil Family}')
    latex_content.append(r'\label{tab:kulfan_coefficients}')
    # latex_content.append(r'\scriptsize')
    
    # Create column specification - Surface | Coeff | thickness columns
    n_thickness_cols = len(sorted_taus)
    col_spec = 'cc' + 'c' * n_thickness_cols
    latex_content.append(r'\begin{tabular}{' + col_spec + '}')
    latex_content.append(r'\toprule')
    
    # Create first header row with "Thickness" spanning over thickness columns
    header_row1 = ['', '', f'\\multicolumn{{{n_thickness_cols}}}{{c}}{{Thickness}}']
    latex_content.append(' & '.join(header_row1) + r' \\')
    latex_content.append(r'\cmidrule(lr){3-' + str(2 + n_thickness_cols) + '}')
    
    # Create second header row with thickness percentages (magnitude only)
    header_row2 = ['Surface', 'Coefficient']
    for tau in sorted_taus:
        header_row2.append(f'{float(tau):.0f}' + '\\%')
    
    latex_content.append(' & '.join(header_row2) + r' \\')
    latex_content.append(r'\midrule')
    
    # Add upper surface coefficients
    for i in range(max_upper):
        if i == 0:
            # First row of upper coefficients - add merged cell
            row = [f'\\multirow{{{max_upper}}}{{*}}{{Upper}}', f'$U_{{{i+1}}}$']
        else:
            row = ['', f'$U_{{{i+1}}}$']
        
        # Add coefficient values for each thickness
        for tau in sorted_taus:
            upper_coeffs = airfoil_data[tau]['upper_coeffs']
            if i < len(upper_coeffs):
                row.append(f"{upper_coeffs[i].magnitude:.4f}")
            else:
                row.append('--')
        
        latex_content.append(' & '.join(row) + r' \\')
    
    latex_content.append(r'\midrule')
    
    # Add lower surface coefficients
    for i in range(max_lower):
        if i == 0:
            # First row of lower coefficients - add merged cell
            row = [f'\\multirow{{{max_lower}}}{{*}}{{Lower}}', f'$L_{{{i+1}}}$']
        else:
            row = ['', f'$L_{{{i+1}}}$']
        
        # Add coefficient values for each thickness
        for tau in sorted_taus:
            lower_coeffs = airfoil_data[tau]['lower_coeffs']
            if i < len(lower_coeffs):
                row.append(f"{lower_coeffs[i].magnitude:.4f}")
            else:
                row.append('--')
        
        latex_content.append(' & '.join(row) + r' \\')
    
    latex_content.append(r'\midrule')
    
    # Add TE gap row
    row = ['\\multicolumn{2}{c}{TE Gap}']
    for tau in sorted_taus:
        row.append(f"{airfoil_data[tau]['te_gap']:.5f}")
    
    latex_content.append(' & '.join(row) + r' \\')
    
    latex_content.append(r'\bottomrule')
    latex_content.append(r'\end{tabular}')
    latex_content.append(r'\end{table}')
    
    # Join all content
    latex_table = '\n'.join(latex_content)
    
    # Write to file
    output_filename = '/Users/codykarcher/Dropbox/publications/2026/01_oso_airfoils/kulfan_coefficients_table.tex'
    with open(output_filename, 'w') as f:
        f.write(latex_table)
    
    return latex_table, airfoil_data

# Generate the table
latex_table, coeffs_data = create_kulfan_coefficients_table()
print(latex_table)
```

```python
def create_airfoil_coordinates_table():
    """
    Extract airfoil coordinates from airfoil files and create a LaTeX table
    """
    # Dictionary to store coordinates for each airfoil
    airfoil_coords = {}
    
    # Process each airfoil file
    for f in files:
        if 'OSO_2025_WT2' in f:
            fnm = f.split('.')
            tau = fnm[0].split('_')[-1][1:]  # Extract thickness from filename
            if float(tau) >= 21:
                # Load the airfoil
                afl = Kulfan(TE_gap=te_gap_lookup[tau])
                afl.readFile(f)
                
                # Store coordinates
                airfoil_coords[tau] = {
                    'thickness': float(tau),
                    'x_coords': afl.xcoordinates,
                    'y_coords': afl.ycoordinates
                }
    
    # Sort by thickness
    sorted_taus = sorted(airfoil_coords.keys(), key=lambda x: float(x))
    
    # Find total number of points (use the first airfoil as reference)
    ref_tau = sorted_taus[0]
    n_points = len(airfoil_coords[ref_tau]['x_coords'])
    
    # Use all points (no sampling)
    all_indices = range(0, n_points)
    
    # Create LaTeX longtable (allows page breaks)
    latex_content = []
    latex_content.append(r'{\renewcommand{\arraystretch}{0.6}')  # Reduce row height
    latex_content.append(r'\begin{longtable}[c]{' + 'c' + 'c' * len(sorted_taus) + '}')
    
    # Caption and label
    latex_content.append(r'\caption{OSO 2025 WT2 Airfoil Coordinates (y/c)} \label{tab:airfoil_coordinates} \\')
    latex_content.append(r'\toprule')
    
    # Create first header row with "Thickness" spanning over thickness columns
    n_thickness_cols = len(sorted_taus)
    header_row1 = ['', f'\\multicolumn{{{n_thickness_cols}}}{{c}}{{Thickness}}']
    latex_content.append(' & '.join(header_row1) + r' \\')
    latex_content.append(r'\cmidrule(lr){2-' + str(1 + n_thickness_cols) + '}')
    
    # Create second header row with thickness percentages
    header_row2 = ['x/c']
    for tau in sorted_taus:
        header_row2.append(f'{float(tau):.0f}' + '\\%')
    
    latex_content.append(' & '.join(header_row2) + r' \\')
    latex_content.append(r'\midrule')
    latex_content.append(r'\endfirsthead')
    
    # Continued header for subsequent pages
    latex_content.append(r'\multicolumn{' + str(1 + n_thickness_cols) + r'}{c}{\tablename\ \thetable\ -- \textit{Continued from previous page}} \\')
    latex_content.append(r'\toprule')
    latex_content.append(' & '.join(header_row1) + r' \\')
    latex_content.append(r'\cmidrule(lr){2-' + str(1 + n_thickness_cols) + '}')
    latex_content.append(' & '.join(header_row2) + r' \\')
    latex_content.append(r'\midrule')
    latex_content.append(r'\endhead')
    
    # Footer for all pages except last
    latex_content.append(r'\midrule')
    latex_content.append(r'\multicolumn{' + str(1 + n_thickness_cols) + r'}{r}{\textit{Continued on next page}} \\')
    latex_content.append(r'\endfoot')
    
    # Footer for last page
    latex_content.append(r'\bottomrule')
    latex_content.append(r'\endlastfoot')
    
    # Add coordinate rows for all points
    for idx in all_indices:
        # Use x-coordinate from reference airfoil (they should be similar)
        x_coord = airfoil_coords[ref_tau]['x_coords'][idx]
        row = [f'{x_coord:.4f}']
        
        # Add y-coordinates for each thickness
        for tau in sorted_taus:
            if idx < len(airfoil_coords[tau]['y_coords']):
                y_coord = airfoil_coords[tau]['y_coords'][idx]
                row.append(f'{y_coord:.4f}')
            else:
                row.append('--')
        
        latex_content.append(' & '.join(row) + r' \\')
    
    latex_content.append(r'\end{longtable}}')  # Close the arraystretch group
    
    # Join all content
    latex_table = '\n'.join(latex_content)
    
    # Write to file
    output_filename = '/Users/codykarcher/Dropbox/publications/2026/01_oso_airfoils/airfoil_coordinates_table.tex'
    with open(output_filename, 'w') as f:
        f.write(latex_table)
    
    print(f"Airfoil coordinates table written to: {output_filename}")
    print(f"Table includes ALL {len(all_indices)} coordinate points")
    print(f"Processed {len(sorted_taus)} airfoils: {', '.join(sorted_taus)}%")
    print("Note: Uses longtable format with reduced row spacing")
    
    return latex_table, airfoil_coords

# Generate the coordinates table
coords_table, coords_data = create_airfoil_coordinates_table()
print(coords_table)
```
